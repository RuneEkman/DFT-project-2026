# $MnI_2$

```python
from magnetic_symmetry_analysis import analyse_magnetic_symmetry
from spinspiral import construct_full
import numpy as np
```

```python
# ─────────────────────────────────────────────────────────────────────────────
# Build supercell
# ─────────────────────────────────────────────────────────────────────────────

theta_mni2, phi_mni2 = 0, 0
path_to_cif_mni2 = "1MnI2-1.cif"
Q_mni2 = [1/3, 1/3, 0]
magnetic_magnitude_mni2 = 4.5

P_mni2 = np.array([
    [2,  1, 0],
    [-1, 1, 0],
    [0,  0, 1]
])

supercell_mni2, name_mni2 = construct_full(
    theta=theta_mni2, phi=phi_mni2, Q=Q_mni2,
    path=path_to_cif_mni2, transform=P_mni2,
    magnitude=magnetic_magnitude_mni2,
    magsymbols='Mn',
    init_moment=[0, magnetic_magnitude_mni2, 0]
)

# ─────────────────────────────────────────────────────────────────────────────
# Run analysis
# ─────────────────────────────────────────────────────────────────────────────

# analyse_magnetic_symmetry(supercell, save_txt=True)
analyse_magnetic_symmetry(supercell_mni2, theta=theta_mni2, phi=phi_mni2, save_txt=False)
```

# $\mathrm{VAl_2Se_4}$

```python
analyse_magnetic_symmetry(supercell_V,theta=theta_V, phi=phi_V, save_txt=False)
```

Forstår ikke helt hvorfor der er så få symmetrier for denne, men odd-parity er bevaret...

Ah, det er fordi SOC er utroligt svag. Den er ikke tvunget til bevaret, SOC er bare for lille.

Den ene symmetri der er tilstede med SOC er $C2[1,-1,0]+TR$ som effektivt set gør følgende på spins:
$ (s_x,s_y,s_z) \rightarrow (s_y,s_x,s_z) $ (swap x og y)

og følgende på space:
$ (x,y,z) \rightarrow (-y,-x,-z)  $

Det er lidt svært at se hvilke sites der bliver mapped til hvilke, og hvorfor det derfor er en symmetri.
